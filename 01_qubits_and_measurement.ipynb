# My first qubit

I start with one wire and two questions. What state did I prepare? What might I see if I measure it?

A fresh qubit starts in |0⟩. Its two amplitudes tell me how much of |0⟩ and |1⟩ is present. Squaring each amplitude’s magnitude gives its measurement probability. The probabilities add up to one.

```python
from math import pi

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

## A coin flip from one gate

H is the Hadamard gate. On |0⟩ it gives equal positive amplitudes. Before running this, guess whether 32 shots have to split exactly 16–16.

```python
coin = QuantumCircuit(1)
coin.h(0)
state = Statevector.from_instruction(coin)
print("Amplitudes:", state.data.round(3))
print("Probabilities:", state.probabilities().round(3))
state.seed(7)
print("32 shots:", dict(state.sample_counts(32)))
```

Each shot samples a freshly prepared state. An amplitude near 0.707 means a probability near 0.5. On a device I collect individual outcomes. Here the simulator also lets me peek at the underlying amplitudes.

```python
coin.h(0)
print("After a second H:", Statevector.from_instruction(coin).probabilities().round(3))
```

The second H brings me back to |0⟩. The paths to |1⟩ cancel. Try inserting `coin.z(0)` just before the second H and rerun from the top. Which outcome becomes certain?

## Which wire did I flip?

Qiskit draws q0 at the top and writes q0 on the right of a basis label. Predict the label for this circuit before expanding the output.

```python
pair = QuantumCircuit(2)
pair.x(0)
print(pair.draw(output="text"))
print(Statevector.from_instruction(pair).probabilities_dict())
```

I read 01 as q1 = 0 and q0 = 1. This detail matters as soon as the two wires hold different bits. [IBM's bit-ordering guide](https://quantum.cloud.ibm.com/docs/en/qiskit/bit-ordering)
